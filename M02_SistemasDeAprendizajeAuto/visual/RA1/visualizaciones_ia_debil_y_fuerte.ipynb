# Visualizaciones: IA débil y fuerte

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
import os

OUT = '.'
os.makedirs(OUT, exist_ok=True)

plt.rcParams['font.family'] = 'DejaVu Sans'
plt.rcParams['font.size'] = 11
print('Setup OK')
```

## Gráfico 1 — Programación Tradicional vs Machine Learning

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.patch.set_facecolor('#0d1117')

colors = {'box': '#1f6feb', 'arrow': '#58a6ff', 'text': 'white', 'bg': '#161b22', 'accent': '#f78166'}

def draw_box(ax, x, y, w, h, label, color='#1f6feb', fontsize=10):
    rect = mpatches.FancyBboxPatch((x - w/2, y - h/2), w, h,
        boxstyle='round,pad=0.05', facecolor=color, edgecolor='#58a6ff', linewidth=1.5)
    ax.add_patch(rect)
    ax.text(x, y, label, ha='center', va='center', color='white', fontsize=fontsize, fontweight='bold', wrap=True)

def draw_arrow(ax, x1, x2, y, color='#58a6ff'):
    ax.annotate('', xy=(x2, y), xytext=(x1, y),
        arrowprops=dict(arrowstyle='->', color=color, lw=2))

for ax in axes:
    ax.set_facecolor(colors['bg'])
    ax.set_xlim(0, 10)
    ax.set_ylim(0, 4)
    ax.axis('off')

# --- Tradicional ---
ax = axes[0]
ax.set_title('Programación Tradicional', color='white', fontsize=13, fontweight='bold', pad=10)
draw_box(ax, 1.5, 2, 2.2, 0.8, 'Datos\n(Inputs)', '#1f6feb')
draw_box(ax, 5, 2, 2.2, 0.8, 'Reglas\n(Código)', '#388bfd')
draw_box(ax, 8.5, 2, 2.2, 0.8, 'Resultado\n(Output)', '#f78166')
draw_arrow(ax, 2.6, 3.9, 2)
draw_arrow(ax, 6.1, 7.4, 2)
ax.text(5, 3.5, 'El programador define\ntodos los pasos', ha='center', color='#8b949e', fontsize=9)

# --- Machine Learning ---
ax = axes[1]
ax.set_title('Machine Learning', color='white', fontsize=13, fontweight='bold', pad=10)
draw_box(ax, 1.5, 2.8, 2.2, 0.8, 'Datos\n(Inputs)', '#1f6feb')
draw_box(ax, 1.5, 1.2, 2.2, 0.8, 'Resultados\nesperados', '#238636')
draw_box(ax, 5, 2, 2.2, 0.8, 'Entrenamiento\n(ML)', '#6e40c9')
draw_box(ax, 8.5, 2, 2.2, 0.8, 'Modelo\n(aprende)', '#f78166')
draw_arrow(ax, 2.6, 3.9, 2.8)
draw_arrow(ax, 2.6, 3.9, 1.2)
ax.annotate('', xy=(3.9, 2.0), xytext=(2.6, 1.2),
    arrowprops=dict(arrowstyle='->', color='#58a6ff', lw=1.5))
draw_arrow(ax, 6.1, 7.4, 2)
ax.text(5, 3.5, 'El modelo aprende\nlos pasos por sí solo', ha='center', color='#8b949e', fontsize=9)

plt.tight_layout()
plt.savefig(f'{OUT}/01_tradicional_vs_ml.png', dpi=150, bbox_inches='tight', facecolor=fig.get_facecolor())
plt.show()
print('Guardado: 01_tradicional_vs_ml.png')
```

## Gráfico 2 — Línea de tiempo histórica de la IA

```python
fig, ax = plt.subplots(figsize=(16, 5))
fig.patch.set_facecolor('#0d1117')
ax.set_facecolor('#0d1117')

events = [
    (1763, 'Teoría\nde Bayes', '#388bfd'),
    (1805, 'Mínimos\ncuadrados', '#388bfd'),
    (1943, 'Neurona\nartificial', '#6e40c9'),
    (1957, 'Perceptrón', '#6e40c9'),
    (1967, 'KNN', '#6e40c9'),
    (1974, 'Backprop', '#6e40c9'),
    (1986, 'Deep\nLearning', '#f0883e'),
    (1995, 'SVM', '#f0883e'),
    (2017, 'Transformers\n(Attention)', '#f78166'),
    (2022, 'ChatGPT\n(boom)', '#ff7b72'),
]

years = [e[0] for e in events]
ax.axhline(y=0.5, color='#30363d', linewidth=3, zorder=1)
ax.set_xlim(1750, 2040)
ax.set_ylim(-0.5, 1.8)
ax.axis('off')
ax.set_title('Historia de la Inteligencia Artificial', color='white', fontsize=14, fontweight='bold', pad=15)

for i, (yr, label, color) in enumerate(events):
    up = (i % 2 == 0)
    y_text = 1.4 if up else -0.2
    y_line_end = 0.65 if up else 0.35
    ax.plot([yr, yr], [0.5, y_line_end], color=color, linewidth=1.5, zorder=2)
    ax.plot(yr, 0.5, 'o', color=color, markersize=9, zorder=3)
    ax.text(yr, y_text, f'{yr}\n{label}', ha='center', va='center' if up else 'top',
            color='white', fontsize=8.5, fontweight='bold')

# Era labels
ax.text(1784, 0.5, '◄ Estadística', ha='left', va='bottom', color='#8b949e', fontsize=8)
ax.text(1940, 0.5, 'ML clásico ►', ha='center', va='bottom', color='#8b949e', fontsize=8)
ax.text(2020, 0.5, 'IA Gen ►', ha='center', va='bottom', color='#8b949e', fontsize=8)

plt.tight_layout()
plt.savefig(f'{OUT}/02_historia_ia.png', dpi=150, bbox_inches='tight', facecolor=fig.get_facecolor())
plt.show()
print('Guardado: 02_historia_ia.png')
```

## Gráfico 3 — IA Débil vs IA Fuerte (tabla comparativa visual)

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 7))
fig.patch.set_facecolor('#0d1117')

card_color_weak = '#161b22'
card_color_strong = '#0d1117'

weak_props = [
    ('✅ Existe hoy', 'Toda la IA actual es IA Débil'),
    ('✅ Tareas específicas', 'ChatGPT, Siri, filtros, Tesla…'),
    ('✅ Aprende de datos', 'Necesita re-entrenamiento'),
    ('❌ Sin conciencia', 'No razona, no entiende'),
    ('❌ Sin flexibilidad', 'Solo su dominio de entrenamiento'),
    ('❌ Sesgos posibles', 'GIGO: Garbage In, Garbage Out'),
]

strong_props = [
    ('🔮 Concepto TEÓRICO', 'No existe ningún sistema hoy'),
    ('🔮 Tareas cualquiera', 'Cualquier tarea cognitiva humana'),
    ('🔮 Conciencia propia', 'Autoconciencia y comprensión'),
    ('🔮 Aprende solo', 'Sin re-entrenamiento humano'),
    ('🔮 Razonamiento real', 'No solo patrones estadísticos'),
    ('⚠️  Riesgos éticos', 'Control, responsabilidad, privacidad'),
]

titles = ['IA DÉBIL (Narrow AI)', 'IA FUERTE (AGI) — Teórica']
title_colors = ['#388bfd', '#6e40c9']
props_list = [weak_props, strong_props]
bg_colors = ['#0a1628', '#13001f']

for ax, title, tcolor, props, bgc in zip(axes, titles, title_colors, props_list, bg_colors):
    ax.set_facecolor(bgc)
    ax.set_xlim(0, 10)
    ax.set_ylim(0, len(props) + 1)
    ax.axis('off')
    ax.set_title(title, color=tcolor, fontsize=13, fontweight='bold', pad=10)
    
    for i, (key, val) in enumerate(reversed(props)):
        y = i + 0.5
        rect = mpatches.FancyBboxPatch((0.2, y + 0.05), 9.6, 0.8,
            boxstyle='round,pad=0.05', facecolor='#161b22', edgecolor=tcolor, linewidth=0.8, alpha=0.7)
        ax.add_patch(rect)
        ax.text(0.5, y + 0.45, key, ha='left', va='center', color='white', fontsize=10, fontweight='bold')
        ax.text(0.5, y + 0.15, val, ha='left', va='center', color='#8b949e', fontsize=9)

plt.suptitle('IA Débil vs IA Fuerte', color='white', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig(f'{OUT}/03_debil_vs_fuerte.png', dpi=150, bbox_inches='tight', facecolor=fig.get_facecolor())
plt.show()
print('Guardado: 03_debil_vs_fuerte.png')
```
